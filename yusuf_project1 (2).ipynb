### Top 10 Best Selling Food Items

This chart represents the top 10 food items based on total quantity sold.

The analysis shows that Chicken Biryani is the highest-selling item with 522 units, followed by Butter Naan with 501 units and Masala Chai with 463 units.

**Business Insight:** The restaurant should maintain sufficient stock of these popular food items to meet customer demand.


```python
# top 10 selling food items

top_items = df.groupby('Menu_Item')['Quantity'].sum().sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_items.values, y=top_items.index)

plt.title('Top 10 Best Selling Food Items')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Food Items')
plt.show()
```

### Monthly Revenue Trend

This line chart shows how restaurant revenue changes month by month across the years 2024 and 2025.

It helps identify revenue fluctuations, high-performing months and periods when sales may be lower.

**Business Insight:** The restaurant can use monthly revenue patterns to plan promotions, inventory and business activities.

```python
# monthly revenue
monthly_sales = df.groupby(
    ['Year', 'Month_Name']
)['Revenue'].sum().reset_index()

monthly_sales['Month_Date'] = pd.to_datetime(
    monthly_sales['Month_Name'] + ' ' +
    monthly_sales['Year'].astype(str),
    format='%B %Y'
)

monthly_sales = monthly_sales.sort_values('Month_Date')

plt.figure(figsize=(14,6))
sns.lineplot(
    data=monthly_sales,
    x='Month_Date',
    y='Revenue',
    marker='o'
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (INR)')
plt.xticks(rotation=45)
plt.show()
```

### Peak Ordering Hours

This chart represents the number of orders received during different hours of the day.

According to the analysis, 9 PM (21:00) is the busiest hour with 343 orders, followed by 8 PM with 284 orders and 3 PM with 283 orders.

**Business Insight:** The restaurant should ensure sufficient staff and food availability during peak ordering hours.

```python
# order hours
hourly_orders = df.groupby('Order_Hour')['Order_ID'].count()

plt.figure(figsize=(12,6))
sns.barplot(x=hourly_orders.index, y=hourly_orders.values)

plt.title('Peak Ordering Hours')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Orders')
plt.show()
```

### Seasonal Food Demand

This chart compares the total quantity of food items sold across different seasons.

The analysis shows:
- Summer: 1,622 units
- Monsoon: 1,343 units
- Winter: 1,249 units
- Spring: 890 units

Summer has the highest demand, while Spring has the lowest demand.

**Business Insight:** The restaurant can plan its inventory and seasonal promotions according to customer demand.

```python
# season demand
seasonal_sales = df.groupby('Season')['Quantity'].sum()

plt.figure(figsize=(9,5))
sns.barplot(x=seasonal_sales.index, y=seasonal_sales.values)

plt.title('Seasonal Food Demand')
plt.xlabel('Season')
plt.ylabel('Quantity Sold')
plt.show()
```

### Discount vs Revenue Analysis

This scatter plot analyzes the relationship between discount percentages and revenue for completed orders.

Each point represents an order. The chart helps explore how different discount levels are associated with order revenue.

**Business Insight:** The restaurant can evaluate discount strategies and identify whether promotions are associated with higher-value orders.

Note: This chart shows association, not proof that discounts cause revenue to increase.

```python
# discount vs revenue

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df[df['Order_Status'] == 'Completed'],
    x='Discount_Percent',
    y='Revenue'
)

plt.title('Discount vs Revenue')
plt.xlabel('Discount Percentage')
plt.ylabel('Revenue (INR)')
plt.show()
```

### Customer Rating Distribution

This histogram shows the distribution of customer ratings given to restaurant orders.

It helps understand how customers rate their experience and whether ratings are concentrated around particular score ranges.

**Business Insight:** The restaurant can monitor customer satisfaction and identify opportunities to improve food quality and service.

```python
# customer rating

plt.figure(figsize=(9,5))

sns.histplot(
    data=df,
    x='Customer_Rating',
    bins=10,
    kde=True
)

plt.title('Customer Rating Distribution')
plt.xlabel('Customer Rating')
plt.show()
```

### Payment Method Distribution

This pie chart represents the distribution of payment methods used by customers.

The dataset includes payment methods such as Cash, UPI, Credit Card and other available payment options.

The project summary identifies Cash as the most popular payment method.

**Business Insight:** The restaurant should provide convenient payment options based on customer preferences.

```python
# payment method
payment = df['Payment_Method'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(
    payment.values,
    labels=payment.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Payment Method Distribution')
plt.show()
```

### Repeat vs New Customer Analysis

This chart compares unique customer IDs grouped by the repeat customer flag.

The analysis helps understand the customer mix and explore customer retention patterns.

**Business Insight:** The restaurant can introduce loyalty programs and customer engagement strategies to encourage repeat visits.

Note: A customer's repeat flag is recorded at order level, so this chart should be interpreted as a grouping by recorded flag, not a verified count of distinct new and returning people.

```python
# customer retaintion

customers = df.groupby(
    'Is_Repeat_Customer'
)['Customer_ID'].nunique()

plt.figure(figsize=(8,5))
sns.barplot(x=customers.index, y=customers.values)

plt.title('Repeat vs New Customers')
plt.xlabel('Customer Type')
plt.ylabel('Unique Customers')
plt.show()
```

### Category-wise Revenue Analysis

This analysis calculates the total revenue generated by each food category.

The results show that Main Course generates the highest revenue at ₹294,676, followed by Fast Food at ₹116,519 and Chinese at ₹108,674.

Bread generates the lowest revenue at ₹20,344.75.

**Business Insight:** Main Course is the leading revenue-generating category in this dataset. The restaurant can focus on maintaining its quality and availability.

```python
# category wise revenue
category_revenue = (
    df.groupby('Category')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
sns.barplot(x=category_revenue.index, y=category_revenue.values)

plt.title('Category-wise Revenue')
plt.xlabel('Food Category')
plt.ylabel('Revenue (INR)')
plt.xticks(rotation=45)
plt.show()
```

```python
# category wise quantity vs revenue
category_analysis = df.groupby('Category').agg(
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('Revenue', 'sum')
).reset_index()

fig, ax1 = plt.subplots(figsize=(10, 5))

sns.barplot(data=category_analysis, x='Category',
            y='Total_Quantity', ax=ax1)

ax1.set_title('Category-wise Quantity Sold')
plt.xticks(rotation=45)
plt.show()
```
